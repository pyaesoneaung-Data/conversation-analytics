# 02 Keyword Search

Apply rule-based product, concern, campaign, and intent classification. No machine learning is used.

## Imports

```python
from pathlib import Path
import sys

import pandas as pd

project_root = Path.cwd().parent
sys.path.insert(0, str(project_root))

from src.preprocess import load_and_preprocess
from src.keyword_search import classify_messages
```

## Configuration

```python
raw_path = project_root / "data" / "raw" / "chatlog_translated_sampled.csv" 
```

## Load Cleaned Data

```python
raw_df, cleaned_df = load_and_preprocess(raw_path)
customer_df = cleaned_df[cleaned_df["standardized_role"] == "customer"].copy()

print("Cleaned messages:", len(cleaned_df))
print("Customer messages:", len(customer_df))
```

## Apply Keyword Classification

```python
classified_df = classify_messages(cleaned_df)

classification_columns = [
    "industry_category", "product_category", "product_subcategory",
    "matched_product_keywords", "primary_concern", "secondary_concerns",
    "matched_concern_keywords", "customer_intent", "intent_confidence_rule",
    "is_campaign_related", "campaign_question_type", "campaign_joining_intent"
]

classified_df[classification_columns].head()
```

## Classification Summary

```python
customer_classified = classified_df[classified_df["standardized_role"] == "customer"].copy()

summary = {
    "top_industries": customer_classified["industry_category"].value_counts().head(10),
    "top_products": customer_classified["product_subcategory"].value_counts().head(10),
    "top_concerns": customer_classified["primary_concern"].value_counts().head(10),
    "top_intents": customer_classified["customer_intent"].value_counts().head(10),
    "campaign_question_types": customer_classified["campaign_question_type"].value_counts().head(10),
}

summary
```
